```python
def sum(a, b):
    return a + b

def subtract(a, b): 
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a /
```

```python
revenue = 1769.98
cost = 250
profit = revenue - cost


# Print
print("The profit was " + str(profit))
```

```python
# Format
print("The profit was U${}. The revenue was U${} and the cost was U${}.".format(profit, revenue, cost))

# F-String
print(f"The profit was U${profit}. The revenue was U${revenue} and the cost was U${cost}.")
```

```python
# Simple BMI Calculation
weight = float(input("Insert your weight"))
height = float(input("Insert your height"))

bmi = weight / (height * height)

print(f"For {weight}kg and {height}m, your BMI is {bmi}")
```
